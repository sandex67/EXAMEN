7. **Tamaño de las compras.** ¿Cuántos tickets corresponden a compras de **hasta 7 unidades**, **de 8 a 14 unidades** y **de 15 o más unidades**? Sume primero las cantidades de todas las líneas de cada ticket.


```python
unidades_por_ticket = df_ventas.groupby('id_ticket')['cantidad'].sum().reset_index()

def clasificar_tamaño(cant):
    if cant <= 7:
        return 'Hasta 7 unidades'
    elif 8 <= cant <= 14:
        return '8 a 14 unidades'
    else:
        return '15 o más unidades'

unidades_por_ticket['tamaño_compra'] = unidades_por_ticket['cantidad'].apply(clasificar_tamaño)

conteo_tamaño = unidades_por_ticket['tamaño_compra'].value_counts().reindex([
    'Tamaño Compra', 'Hasta 7 unidades', '8 a 14 unidades', '15 o más unidades'
]).rename('Cantidad de tickets').to_frame()

display(conteo_tamaño)
```

**Justificación del procedimiento**

Para resolver este ejercicio, utilicé el archivo `ventas.csv`, ya que contenía el identificador de cada ticket y la cantidad de productos vendidos.

Primero, agrupé los registros mediante la columna `id_ticket` y sumé las cantidades de todas las líneas correspondientes a cada ticket. Para esto utilicé los métodos `groupby()` y `sum()`, lo que me permitió conocer el total de unidades compradas en cada transacción.

Posteriormente, creé una función llamada `clasificar_tamano()` que me permitió clasificar cada ticket según la cantidad total de unidades compradas. Establecí tres categorías: hasta 7 unidades, de 8 a 14 unidades y 15 o más unidades.

Finalmente, utilicé `value_counts()` para contar cuántos tickets pertenecían a cada categoría y presenté los resultados en una tabla.

**Explicación del código**

1. `groupby('id_ticket')`: agrupé los registros que pertenecían al mismo ticket.
2. `['cantidad']`: seleccioné la columna que contenía las unidades vendidas.
3. `sum()`: sumé las cantidades de todas las líneas de cada ticket para conocer cuántas unidades se compraron en total.
4. `reset_index()`: convertí el resultado agrupado nuevamente en un DataFrame.
5. `def clasificar_tamano(cant)`: creé una función que recibía la cantidad total de unidades de cada ticket para determinar su tamaño.
6. `if cant <= 7`: establecí la primera condición para clasificar las compras de hasta 7 unidades.
7. `return 'Hasta 7 unidades'`: devolví el nombre de la primera categoría cuando se cumplía esa condición.
8. `elif 8 <= cant <= 14`: establecí la segunda condición para identificar las compras que tenían entre 8 y 14 unidades.
9. `return '8 a 14 unidades'`: devolví el nombre de la segunda categoría cuando se cumplía esa condición.
10. `else`: establecí una última condición para los tickets que no pertenecían a las dos categorías anteriores.
11. `return '15 o más unidades'`: clasifiqué en esta categoría los tickets que contenían 15 o más unidades.
12. `apply(clasificar_tamano)`: apliqué la función a las cantidades de todos los tickets para clasificarlos individualmente.
13. `value_counts()`: conté cuántos tickets pertenecían a cada categoría.
14. `reindex()`: organicé las categorías en el orden solicitado por el ejercicio.
15. `rename('Cantidad de tickets')`: asigné un nombre descriptivo a la columna de resultados.
16. `to_frame()`: convertí los resultados en una tabla.
17. `display()`: presenté la tabla con la cantidad de tickets de cada categoría.

8. **Ventas por mes.** ¿Cuántos tickets y qué importe total se registraron en cada mes? ¿En qué mes se alcanzó el mayor importe?


```python
df_ventas['fecha'] = pd.to_datetime(df_ventas['fecha'])

df_ventas['importe_linea'] = df_ventas['cantidad'] * df_ventas['precio_unitario']

ventas_mes = df_ventas.groupby(df_ventas['fecha'].dt.to_period('M')).agg({
    'id_ticket': 'nunique',
    'importe_linea': 'sum'
}).rename(columns={
    'id_ticket': 'Cantidad de tickets',
    'importe_linea': 'Importe Total ($)'
}).round(2)

display(ventas_mes)

mayor_mes = ventas_mes['Importe Total ($)'].idxmax()

print("El mes con mayor importe de ventas fue:", mayor_mes)
                                    
```

**Justificación del procedimiento**

Posteriormente, agrupé los registros por mes para obtener dos datos: la cantidad de tickets distintos y el importe total de las ventas. Finalmente, identifiqué el mes que alcanzó el mayor importe utilizando `idxmax()`.

**Explicación del código**

1. `pd.to_datetime()`: convertí la columna de fechas a un formato de fecha que pandas podía reconocer y utilizar para agrupar los registros.
2. `df_ventas['fecha']`: seleccioné la columna que contenía las fechas de las ventas.
3. `importe_linea`: creé una columna para almacenar el importe correspondiente a cada línea de venta.
4. `cantidad * precio_unitario`: calculé el importe de cada línea multiplicando las unidades vendidas por el precio de cada unidad.
5. `groupby()`: agrupé los registros de ventas según el mes al que pertenecía cada fecha.
6. `dt.to_period('M')`: extraje el período mensual de cada fecha, considerando tanto el año como el mes.
7. `agg()`: apliqué diferentes operaciones de cálculo a las columnas que necesitaba analizar.
8. `'id_ticket': 'nunique'`: conté los tickets distintos de cada mes, evitando contar varias veces un mismo ticket.
9. `'importe_linea': 'sum'`: sumé los importes de todas las líneas de venta correspondientes a cada mes.
10. `rename()`: cambié los nombres de las columnas para identificar claramente la cantidad de tickets y el importe total.
11. `round(2)`: redondeé los importes a dos decimales.
12. `display()`: presenté la tabla con los resultados mensuales.
13. `idxmax()`: identifiqué el mes que tenía el mayor importe total de ventas.

9. **Producto más vendido por sucursal.** ¿Qué producto acumuló más unidades vendidas en cada sucursal? Indique su código, nombre y cantidad vendida. Si hay un empate, incluya todos los productos empatados.


```python
ventas_productos = pd.merge(
    df_ventas,
    df_productos[['id_producto', 'producto']],
    on='id_producto',
    how='left'
)

unidades_producto = ventas_productos.groupby(
    ['sucursal', 'id_producto', 'producto'],
    as_index=False
)['cantidad'].sum()

maximo_sucursal = unidades_producto.groupby('sucursal')['cantidad'].transform('max')

mas_vendidos = unidades_producto[
    unidades_producto['cantidad'] == maximo_sucursal
].sort_values('sucursal')

display(mas_vendidos)
```

**Justificación del procedimiento**

Para resolver este ejercicio, utilicé dos archivos: `ventas.csv` y `productos.csv`.

El archivo de ventas contenía el código de cada producto vendido, la sucursal y la cantidad de unidades. Por otro lado, el catálogo de productos contenía el código y el nombre de cada producto.

Por esta razón, primero relacioné ambos archivos mediante la columna `id_producto`, utilizando el método `merge()`. Esto me permitió incorporar el nombre de cada producto a los registros de ventas.

Posteriormente, agrupé la información por sucursal, código y nombre del producto, sumando las unidades vendidas de cada uno.

Finalmente, identifiqué la cantidad máxima de unidades vendidas en cada sucursal y seleccioné los productos que alcanzaron ese máximo. De esta manera, también pude incluir todos los productos cuando existió un empate.

**Explicación del código**

1. `pd.merge()`: uní los archivos de ventas y productos para relacionar la información de ambos DataFrames.
2. `df_ventas`: seleccioné el archivo que contenía los registros de las ventas.

10. **Compras por tipo de cliente.** Para cada tipo de cliente, ¿cuántos clientes distintos compraron, cuántos tickets generaron y cuál fue el importe total de sus compras?


```python
df_v_c = pd.merge(df_ventas, df_clientes, left_on='id_cliente', right_on='id', how='inner')

if 'importe_linea' not in df_v_c.columns:
    df_v_c['importe_linea'] = df_v_c['cantidad'] * df_v_c['precio_unitario']

compras_tipo_cliente = (
    df_v_c.groupby('tipo_cliente').agg({'id_cliente': 'nunique', 'id_ticket': 'nunique', 'importe_linea': 'sum'})
    .rename(columns={
        'id_cliente': 'Clientes_Distintos',
        'id_ticket': 'Tickets_Generados',
        'importe_linea': 'Importe_Total_USD'
    }).round(2).sort_index()
)

display(compras_tipo_cliente)
```

**Justificación del procedimiento**

Para resolver este ejercicio, utilicé dos archivos: `ventas.csv` y `clientes.csv`.

El archivo de ventas contenía los registros de las compras realizadas, incluyendo el identificador del cliente, el ticket, la cantidad de productos y el precio unitario. Por otro lado, el archivo de clientes contenía el tipo al que pertenecía cada cliente, como estudiante, docente o personal administrativo.

Por esta razón, primero relacioné ambos archivos mediante el identificador del cliente, utilizando `merge()`. Esto me permitió incorporar el tipo de cliente a los registros de ventas.

Posteriormente, calculé el importe de cada línea de venta multiplicando la cantidad por el precio unitario. Después, agrupé los registros según el tipo de cliente y utilicé `agg()` para obtener tres resultados: clientes distintos, tickets generados e importe total de las compras.

Finalmente, presenté los resultados en una tabla, redondeando los importes a dos decimales.

**Explicación del código**

1. `pd.merge()`: uní los archivos de ventas y clientes para relacionar la información de ambos DataFrames.
2. `df_ventas`: seleccioné el archivo que contenía los registros de las compras realizadas.
